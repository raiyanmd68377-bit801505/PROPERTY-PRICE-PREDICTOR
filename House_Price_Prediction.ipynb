```python



def predict_house_price(
    area, bedrooms, bathrooms, stories, parking,
    mainroad, guestroom, basement, hotwaterheating,
    airconditioning, prefarea, furnishingstatus
):
    data = {
        'area': [area],
        'bedrooms': [bedrooms],
        'bathrooms': [bathrooms],
        'stories': [stories],
        'parking': [parking],

        'mainroad_no': [int(mainroad == 'no')],
        'mainroad_yes': [int(mainroad == 'yes')],

        'guestroom_no': [int(guestroom == 'no')],
        'guestroom_yes': [int(guestroom == 'yes')],

        'basement_no': [int(basement == 'no')],
        'basement_yes': [int(basement == 'yes')],

        'hotwaterheating_no': [int(hotwaterheating == 'no')],
        'hotwaterheating_yes': [int(hotwaterheating == 'yes')],

        'airconditioning_no': [int(airconditioning == 'no')],
        'airconditioning_yes': [int(airconditioning == 'yes')],

        'prefarea_no': [int(prefarea == 'no')],
        'prefarea_yes': [int(prefarea == 'yes')],

        'furnishingstatus_furnished': [
            int(furnishingstatus == 'furnished')
        ],
        'furnishingstatus_semi-furnished': [
            int(furnishingstatus == 'semi-furnished')
        ],
        'furnishingstatus_unfurnished': [
            int(furnishingstatus == 'unfurnished')
        ]
    }

    input_df = pd.DataFrame(data, columns=X_train.columns)

    prediction = loaded_model.predict(input_df)

    return prediction[0]
```

```python
price = predict_house_price(
    area=5000,
    bedrooms=3,
    bathrooms=2,
    stories=2,
    parking=1,
    mainroad='yes',
    guestroom='no',
    basement='yes',
    hotwaterheating='no',
    airconditioning='yes',
    prefarea='yes',
    furnishingstatus='furnished'
)

print("Predicted House Price:", price)
```
